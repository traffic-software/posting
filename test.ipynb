```python
from xml.dom.minidom import Element
import mysql.connector
from selenium import webdriver
import time
```

```python
driver = webdriver.Chrome()

driver.get('https://www.amazon.com/Krylon-K02203-General-Purpose-Metallic/dp/B0042T7TRC/ref=cm_cr_arp_d_product_top?ie=UTF8&th=1')

```

```python
from selenium.webdriver.common.by import By
productsname = driver.find_element(By.XPATH, "//span[@id='productTitle']").text
print(productsname)
```

```python

for ask in driver.find_elements(By.CLASS_NAME, 'askTeaserQuestions'):
   
    
    print(ask.find_element(By.XPATH, '//div[starts-with(@id,"question")]').text)

```
